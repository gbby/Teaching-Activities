```python
# Read in the data (make sure you download each file and place them in a "data" directory where your .ipynb is located)
cities = pd.read_csv('/content/drive/Shareddrives/MBAI 5400G - Visualization and Storytelling (Winter 2026)/In-Class Activities/Python Visualization/data/minard-cities.csv')
troops = pd.read_csv('/content/drive/Shareddrives/MBAI 5400G - Visualization and Storytelling (Winter 2026)/In-Class Activities/Python Visualization/data/minard-troops.csv')
weather = pd.read_csv('/content/drive/Shareddrives/MBAI 5400G - Visualization and Storytelling (Winter 2026)/In-Class Activities/Python Visualization/data/minard-temp.csv')
cities.head()
```

```python
named_city = cities[cities['city'].notna()]
named_city
```

```python
fig = px.scatter_geo(named_city,lat='lat',lon='long',color='direction',
                     hover_name='city',size='survivors',projection='equirectangular',scope='europe',
                     title='Minard map')
fig.show()
```

```python
survivors = px.density_mapbox(troops, lat = 'lat', lon = 'long',
                         z = 'survivors',
                         center = dict(lat = 55, lon = 30),
                         zoom = 4.7,
                         hover_name = 'direction',
                         mapbox_style = 'carto-positron'
                         )
survivors.show()
```
